```python
machines = {
    'm1': '192.168.0.1',
    'm2': '192.168.0.2',
    'm3': '192.168.0.3',
    'm4': '192.168.0.4',
    'm5': '192.168.0.5'
}
# 1
print(machines['m2'])
# 2
print(len(machines))
# 3
machines['m6'] = '192.168.0.6'
print(machines)
# 4
machines.pop('m4')
print(machines)

# 5
print('m2' in machines)

key = input("Entrez le nom d'une machine")
if key in machines:
    print(f"L'adresse IP de machine {key} est {machines[key]}")
else:
    print(f"La machine {key} n'est pas répertoriée.")
```
